```python
from itertools import combinations

pairs = list(
    combinations(cards["title"], 2)
)

len(pairs)
```

```python
pairs_df = pd.DataFrame(
    pairs,
    columns=["card1", "card2"]
)

pairs_df.head()
```

```python
pairs_df = pairs_df.merge(
    cards[["title","theme"]],
    left_on="card1",
    right_on="title"
).rename(columns={"theme":"theme1"}).drop(columns="title")

pairs_df = pairs_df.merge(
    cards[["title","theme"]],
    left_on="card2",
    right_on="title"
).rename(columns={"theme":"theme2"}).drop(columns="title")
```

```python
theme_categories = {

    "Origins": [
        "Seed",
        "Herald",
        "Witness",
        "Growth",
        "Guardian"
    ],

    "Departure": [
        "Exile",
        "Lover",
        "Wanderer",
        "Initiation"
    ],

    "Descent": [
        "Ancestors",
        "Rebirth",
        "Recognition",
        "Perspective",
        "Wisdom"
    ],

    "Restoration": [
        "Healer",
        "Power",
        "Illumination",
        "Hope"
    ],

    "Integration": [
        "Subconscious",
        "Consciousness",
        "Nourishment"
    ],

    "Mastery": [
        "Obstacles Overcome",
        "Responsibility",
        "Overview"
    ],

    "Transformation": [
        "Flood",
        "Transformation",
        "Wholeness"
    ],

    "Sovereignty": [
        "Accomplishment",
        "Sovereignty"
    ]
}
```

```python
rows = []

for category, themes in theme_categories.items():
    for theme in themes:
        rows.append({
            "category": category,
            "theme": theme
        })

theme_df = pd.DataFrame(rows)

theme_df
```

```python
import pandas as pd

cards = pd.read_csv("harpy_cards_meanings.csv")

print(len(cards))
cards.columns.tolist()
```

```python
cards["image"] = cards.apply(
    lambda row:
        f"{int(row['number']):02d}-"
        + row["title"]
            .lower()
            .replace(" ", "-")
            .replace("'", "")
        + ".png",
    axis=1
)

cards[["number","title","image"]].head(10)
```

```python
import json

website_cards = cards[
    [
        "number",
        "title",
        "keywords",
        "oracle",
        "meaning",
        "image"
    ]
].copy()

website_cards["keywords"] = (
    website_cards["keywords"]
    .fillna("")
    .str.split("|")
)

with open(
    "../the-oracle-of-harpies/cards.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        website_cards.to_dict("records"),
        f,
        indent=2,
        ensure_ascii=False
    )

print("Created cards.json with", len(website_cards), "cards")
```
