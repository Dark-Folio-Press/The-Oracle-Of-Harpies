```python
from itertools import combinations

pairs = list(combinations(cards, 2))

print(len(pairs))
```

```python
from itertools import combinations
import pandas as pd

pair_rows = []

for c1, c2 in combinations(cards, 2):
    pair_rows.append({
        "card1": c1["title"],
        "card2": c2["title"]
    })

pairs_df = pd.DataFrame(pair_rows)

print(len(pairs_df))
pairs_df.head()
```

```python
triple_rows = []

for c1, c2, c3 in combinations(cards, 3):
    triple_rows.append({
        "card1": c1["title"],
        "card2": c2["title"],
        "card3": c3["title"]
    })

triples_df = pd.DataFrame(triple_rows)

print(len(triples_df))
triples_df.head()
```

```python
pairs_df["theme"] = ""
pairs_df["oracle"] = ""
pairs_df["meaning"] = ""

pairs_df.to_csv(
    "harpies_pair_interpretations.csv",
    index=False
)
```

```python
triples_df["theme"] = ""
triples_df["oracle"] = ""
triples_df["meaning"] = ""

triples_df.to_csv(
    "harpies_triple_interpretations.csv",
    index=False
)
```
